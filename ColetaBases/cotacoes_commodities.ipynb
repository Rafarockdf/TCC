```python
import yfinance as yf
import pandas as pd

def descobrir_variaveis_agro():
    termos = ["Soybean Futures", "Corn Futures", "Coffee Futures", "Soybean Meal", "USD/BRL"]
    lista_final = []

    for termo in termos:
        # Realiza a busca no banco de dados do Yahoo Finance
        busca = yf.Search(termo, max_results=5)
        
        for resultado in busca.quotes:
            lista_final.append({
                "Descricao_Busca": termo,
                "Ticker": resultado.get('symbol'),
                "Nome": resultado.get('shortname'),
                "Bolsa": resultado.get('exchange'),
                "Tipo": resultado.get('quoteType')
            })

    return pd.DataFrame(lista_final)

# Executa e exibe a tabela de variáveis encontradas
df_variaveis = descobrir_variaveis_agro()
print(df_variaveis)
```
